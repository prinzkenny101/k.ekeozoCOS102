```python
import random

class Employee:
    def __init__(self):
        self.employees = ["Mary Evans", "Eyo Ishan", "Durojaiye Dare", "Adams Ali", "Andrew Ugwu", 
                          "Stella Mankinde", "Jane Akibo", "Ago James", "Michell Taiwo", 
                          "Abraham Jones", "Nicole Anide", "Kosi Korso", "Adele Martins", 
                          "Emmanuel Ojo", "Ajayi Fatima"]
        self.tasks = ["Loading", "Transporting", "Reviewing Orders", "Customer Service", "Delivering Items"]

    def check_employee(self, name):
        if name in self.employees:
            return True
        return False

    def take_attendance(self, name):
       return f"Attendance taken for {name}."

    def assign_task(self):
       return random.choice(self.tasks)

    def refuse_access(self):
       return "Access Refused: You are not a recognized employee."
```

```python
import tkinter as tk
from tkinter import messagebox as msgbox

# main window
root = tk.Tk()
root.title("Mrs Jane's Logistics System")
root.geometry("400x250")

emp_system = Employee()

def run_process():
    user_name = name_entry.get().strip()

    if emp_system.check_employee(user_name):
        attendance = emp_system.take_attendance(user_name)
        task = emp_system.assign_task()

        message = f"Welcome, {user_name}!\n{attendance}\nYour task for today is: {task}"
        msgbox.showinfo("Success", message)
    else:
        error_msg = emp_system.refuse_access()
        msgbox.showerror("Error", error_msg)

# GUI Layout
tk.Label(root, text="Enter Employee Name:").pack(pady=10)
name_entry = tk.Entry(root, width=30)
name_entry.pack(pady=0)

login_button = tk.Button(root, text="Check Identity", command=run_process)
login_button.pack(pady=20)

root.mainloop()
```

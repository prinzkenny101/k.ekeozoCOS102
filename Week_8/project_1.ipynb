```python
import tkinter as tk
from tkinter import messagebox as msgbox

class Zenith:
    def __init__(self, name):
       self.name = name

    def mutual_services(self):
        return ["Lines of credit", "Investment management and accounts", "Insurance"]

    def unique_services(self):
        pass

class RetailBanking(Zenith):
    def unique_services(self):
        return ["Retirement and education accounts", "Loans and mortgages", "Checking and saving"]

class GlobalBanking(Zenith):
    def unique_services(self):
        return ["Multi-currency management services and products", "Foreign currency accounts", "Foreign currency credit cards", "Transborder advisory services", "Liquidity management"]

class CommercialBanking(Zenith):
    def unique_services(self):
        return["Advisory services"]

def show_services():
    name = name_entry.get()
    division = division_var.get()

    if name == "Mary" and division == "Retail Banking":
        employee_obj = RetailBanking(name)
    elif name == "Agatha" and division == "Global Banking":
        employee_obj = GlobalBanking(name)
    elif name == "Noel" and division == "Commercial Banking":
        employee_obj = CommercialBanking(name)
    else:
        msgbox.showerror("Error", "Invalid Name or Division")
        return

    all_services = employee_obj.mutual_services() + employee_obj.unique_services()
    msgbox.showinfo("Services", f"Services for {name}:\n" + "\n".join(all_services))
```

```python
#main window
root = tk.Tk()
root.title("Zenith Bank")
root.geometry("400x250")

tk.Label(root, text="Enter Employee Name:").pack(pady=10)
name_entry = tk.Entry(root, width=30)
name = name_entry.get().strip().capitalize()
name_entry.pack(pady=0)

division_var = tk.StringVar(root)
division_var.set("Select Division")

options = ["Retail Banking", "Global Banking", "Commercial Banking"]

tk.Label(root, text="Select Division:").pack(pady=10)
division_menu = tk.OptionMenu(root, division_var, *options)
division_menu.pack(pady=5)

submit_button = tk.Button(root, text="Check Services", command=show_services)
submit_button.pack(pady=20)

root.mainloop()
```

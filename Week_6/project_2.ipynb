```python
import tkinter as tk
from tkinter import messagebox as msgbox 

class Delivery:
    def calculate_price(self, location, weight):
        location = location.strip().upper()

        if location == "PAU":
            if weight >= 10:
                return "Your amount is N2000"
            else:
                return "Your amount is N1500"

        elif location == "Epe":
            if weight >= 10:
                return "Your amount is N5000"
            else:
                return "Your amount is N4000"
                
        else: 
            return "Sorry, we don't deliver packages to this place at the moment"

# Setting up the GUI

def get_quote():
    try:
        location_input = location_entry.get()
        weight_input = float(weight_entry.get())

        service = Delivery()
        result = service.calculate_price(location_input, weight_input)

        if "N" in result: 
            msgbox.showinfo("Quote", result)
        else:
            msgbox.showwarning("Not Available", result)

    except ValueError:
        msgbox.showerror("Error, Please enter a valid number for weight.")
```

```python
#main window
root = tk.Tk()
root.title("Kenny's Delivery service")
root.geometry("400x300")

# Widgets
tk.Label(root, text="Enter Location (PAU or Epe):").pack(pady=10)
location_entry = tk.Entry(root)
location_entry.pack()

tk.Label(root, text="Enter Weight (kg):").pack(pady=10)
weight_entry = tk.Entry(root)
weight_entry.pack()

button = tk.Button(root, text="Calculate Cost", command=get_quote)
button.pack(pady=20)

root.mainloop()
```
